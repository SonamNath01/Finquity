write the code toperform basic text processing task the program should create a corpus using user defined word string or text file and egenerate a vocabulary from the corpus it should then clean the corpus for removing all stop words after cleaning the program must save the generated vocabulry in json format and subsequently load the vocabulary for the further use

```python
import nltk
import json
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download required resources
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
def create_corpus_from_text(text):
    tokens = word_tokenize(text.lower())
    return tokens
```

```python
# Create corpus from text file
def create_corpus_from_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        text = file.read().lower()
    tokens = word_tokenize(text)
    return tokens
```

```python
# Clean corpus (remove stopwords & punctuation)
def clean_corpus(tokens):
    stop_words = set(stopwords.words('english'))
    cleaned_tokens = [
        word for word in tokens
        if word not in stop_words and word not in string.punctuation
    ]
    return cleaned_tokens
```

```python
# Generate vocabulary
def generate_vocabulary(tokens):
    vocabulary = sorted(list(set(tokens)))
    return vocabulary
```

```python
# Save vocabulary to JSON
def save_vocabulary_to_json(vocabulary, file_path):
    with open(file_path, 'w') as json_file:
        json.dump(vocabulary, json_file, indent=4)
    print(f"Vocabulary saved to {file_path}")
```

```python

# Load vocabulary from JSON
def load_vocab_from_json(file_path):
    with open(file_path, 'r') as json_file:
        vocab = json.load(json_file)
    print("Vocabulary loaded from JSON file")
    return vocab
```

```python
text = "Natural Language Processing is very interesting and useful."

corpus = create_corpus_from_text(text)
cleaned = clean_corpus(corpus)
vocab = generate_vocabulary(cleaned)

save_vocabulary_to_json(vocab, "vocabulary.json")
loaded_vocab = load_vocab_from_json("vocabulary.json")

print("Final Vocabulary:", loaded_vocab)
```
